```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("images (3).jpg" , cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap = 'gray')
```

```python
def add(img,prob=0.1):
  n = img.copy()
  ran = np.random.rand(*img.shape)
  n[ran < prob/2] = 0
  n[ran > 1-prob/2] = 255
  return n
```

```python
img1 = add(img,0.2)
plt.imshow(img1, cmap = 'gray')
```

```python
mean = img1.mean()
print(mean)
def remove(img, mean = 0.5):
  m = img.copy()
  m[m == 255] = int(mean)
  m[m == 0] = int(mean)
  return m
```

```python
img_new = remove(img,mean)
plt.imshow(img_new, cmap = 'gray')
```

```python
import cv2
import numpy as np
img1 = cv2.imread("images (3).jpg")
med = np.median(img1)
print(med)
def remove1(img,med= 0.5):
  m = img.copy()
  m[m == 255] = int(med)
  m[m == 0] = int(med)
  return m
```

```python
img_ = img.copy()
for i in range(img1.shape[0]):
  for j in range(img1.shape[1]):
    i1 = img[i:i+3,j:j+3]
    img_[i,j] = np.median(i1)
plt.imshow(img_,cmap = 'gray')
```

```python
img_1 = img.copy()
for i in range(img1.shape[0]-2):
  for j in range(img1.shape[1]-2):
    i1 = img[i:i+3,j:j+3]
    i1[i+1,j+1] = np.mean(i1)
plt.imshow(img_1,cmap = 'gray')
```
